2.8. Уравнения неразрывности

Рассмотрим материальную частицу, которая в момент $t=0$ находилась в точке с координатами $(х_1, х_2, х_3)$ в прямоугольной декартовой системе координат. По прошествии времени $t$ эта частица переместится в другую точку с координатами $(\xi_1, \xi_2, \xi_3)$ относительно той же самой системы координат. Зависимость

$$\xi_i = \xi_i (x_1, x_2, x_3, t), i = 1,2,3,$$

описывает конфигурацию тела в разные моменты времени $t$. Функции $\xi_i$ должны быть непрерывными однозначными функциями с отличным от нуля якобианом.

Рассмотрим скорость и ускорение материальной частицы, траектория которой задана соотношениями (1). В описании Лагранжа имеем

$$v_i(\vec x, t) = \frac{\partial}{\partial t} \xi_i (\vec x, t),$$

$$\dot v_i(\vec x, t) = \frac{\partial^2}{\partial t^2} \xi_i (\vec x, t) = \frac{\partial}{\partial t} v_i (\vec x, t).$$

В этих выражениях дифференцирование по времени производится при постоянных значениях $\vec x \equiv (x_1, x_2, x_3)$.

В описании Эйлера в качестве независимых переменных принимаются координаты $\xi$ и время $t$. В этом описании движение тела представляется векторным полем $v_i(\xi_1, \xi_2, \xi_3, t)$, связанным с мгновенным положением материальной частицы. Для вычисления ускорения материальной частицы рассмотрим эту частицу в двух близких положениях. Частица, в момент $t$ занимающая положение $\xi_i$, перемещается так, что в следующий момент $t+dt$ она занимает положение $\xi_i + v_i dt$.

Ускорение $\dot v_i(\vec \xi, t)$ можно представить как разность скоростей

$$\dot v_i(\xi_j, t) = \frac{v_i(\xi_j+v_j dt, t+dt)-v_i(\xi_j, t)}{dt}$$

Разлагая первый член правой части этого уравнения в ряд Тейлора и ограничиваясь величинами первого порядка малости по $dt$, имеем

$$\dot v_i(\xi_j, t) = \frac{D v_i(\xi_j, t)}{Dt} = \frac{\partial}{\partial t} v_i(\xi_j, t) + v_j \frac{\partial v_i(\vec \xi, t)}{\partial \xi_j}$$

Первый член правой части уравнения (5) описывает изменение во времени функции $v_i$ при постоянном $\xi$ т. е. при неизменном положении в пространстве, а не для одной и той же материальной частицы. Чтобы подсчитать изменение скорости $v_i$ данной частицы, нужно в течение некоторого времени проследить за ее движением. К величине $\frac{\partial v_i(\xi_j, t)}{\partial t} \Big|_{\vec \xi = const}$ требуется добавить член $v_j \frac{\partial v_i(\vec \xi, t)}{\partial \xi_j}$ как вклад движения частицы в мгновенное поле скоростей. Производная $\dot v_i(\xi_j, t)$ называется <i>местной производной</i> или
<i>материальной производной</i>. Член $v_j \frac{\partial v_i(\vec \xi, t)}{\partial \xi_j}$ называется конвективной производной.

Подобное рассуждение можно повторить для любой другой характеристики материальной среды, например для ее плотности или температуры. Вообще имеем

$$\dot G = \frac{D G(\xi_j, t)}{Dt} = \frac{\partial G(\xi_j, t)}{\partial t}\Big|_{\vec \xi = const}  + v_j \frac{\partial G(\vec \xi, t)}{\partial \xi_j} = \frac{\partial G(\xi_j, t)}{\partial t}\Big|_{x = const}$$

К местной производной относятся все обычные правила дифференцирования, обычные формулы производной от суммы, разности, произведения.

2.9. Уравнения движения

Уравнения движения мы получим из двух основных законов механики: закона изменения количества движения и закона изменения момента количества движения. Составляющие количества движения тела в момент $t$ выражаются формулой

$$P_{i} = \int\limits_{V} \rho(\vec \xi) v_i(\vec \xi, t) dV$$

Закон изменения количества движения в описании Эйлера, т. е. в координатах $\xi_i$, имеет вид

$$\frac{D P_i}{D t} = \int\limits_{V} X_i dV + \int\limits_{A} p^{(n)}_i dA$$

Изменение количества движения во времени равно главному вектору сил, действующих на область $V$. Этот главный вектор равен интегралу массовых сил $Х_i$ по объему тела, а также интегралу поверхностных сил по поверхности тела. Выражая вектор напряжения $p^{(n)}_i$ через компоненты тензора напряжений $\sigma_{ij}$ и применяя теорему Гаусса — Остроградского, имеем

$$\frac{D P_i}{D t} = \int\limits_{V} X_i dV + \int\limits_{A} \sigma_{ji}{n}_j dA =
\int\limits_{V} \left(X_i + \frac{\partial \sigma_{ji}}{\partial \xi_j} \right) dV$$

Дифференцируя формулу (1), имеем

$$\int\limits_{V} \left[\frac{\partial \left(\rho(\vec \xi) v_i(\vec \xi, t)\right)}{\partial t} + \frac{\partial}{\partial \xi_j}\left(\rho(\vec \xi) v_i(\vec \xi, t) v_j(\vec \xi, t)\right)\right] dV =
\int\limits_{V} \left(X_i + \frac{\partial \sigma_{ji}}{\partial \xi_j} \right) dV$$

Здесь мы использовали формулу (13) предыдущего параграфа. Закон изменения количества движения (2) был применен к произвольной области У тела. Если подинтегральное выражение непрерывно, то, в силу произвольности выбранной области, уравнение (4) приводит к локальному соотношению

$$\frac{\partial \left(\rho(\vec \xi) v_i(\vec \xi, t)\right)}{\partial t} + \frac{\partial}{\partial \xi_j}\left(\rho(\vec \xi) v_i(\vec \xi, t) v_j(\vec \xi, t)\right) =
X_i + \frac{\partial \sigma_{ji}}{\partial \xi_j}$$

Уравнение (5) допускает значительное упрощение благодаря уравнению неразрывности

$$\frac{\partial \rho(\vec \xi) }{\partial t} + \frac{\partial}{\partial \xi_j}\left(\rho(\vec \xi) v_j(\vec \xi, t)\right) =
0$$

Учитывая формулу (6), приводим (5) к виду

$$\rho\frac{Dv^i}{Dt} = X_i + \frac{\partial \sigma_{ji}}{\partial \xi_j}$$

Эта формула представляет собой локальную запись закона изменения количества движения.

При выводе уравнений движения мы использовали описание Эйлера, принимая в качестве независимых переменных коордннаты $\xi_i$ и время $t$.

Структура полученных уравнений достаточно проста (уравнения (7) и (15)), несравненно проще, чем в описании Лагранжа. Однако описание Эйлера в некоторых отношениях неудобно. В задачах нелинейной теории упругости, как правило, известны первоначальные положения точек и разыскиваются поля перемещений, вызванные деформацией тела. Граничные условия в виде заданных нагрузок или перемещений также просто выражаются в координатах $x_i$ Неудобством является и то, что дифференцирование в уравнениях (7) производится относительно переменных $\xi_i$, содержащих разыскиваемые величины — перемещения.

Из уравнений движения в описании Эйлера легко перейти к уравнениям движения линейной теории упругости, в которой рассматриваются малые деформации и малые скорости перемещений. Линеаризуем прежде всего выражение для тензора деформаций. Пренебрежем в тензоре Альманси

$$\eta_{ij} = \frac{1}{2}\left(\frac{\partial u_i}{\partial \xi_j}+\frac{\partial u_j}{\partial \xi_i} - \frac{\partial u_k}{\partial \xi_i}\frac{\partial u_k}{\partial \xi_j}\right)$$

нелинейным членом. Так как при малых деформациях производные по переменным $x_i$ и $\xi_i$ будут различаться на величины высшего порядка малости, для тензора деформаций примем выражение

$$\epsilon_{ij} = \frac{1}{2}\left(\frac{\partial u_i}{\partial \xi_j}+\frac{\partial u_j}{\partial \xi_i}\right)$$

Переходя к уравнениям движения (7), заметим, что при малых деформациях и малых скоростях перемещений нелинейные члены в местных производных пренебрежимы по сравнению с материальными производными. Таким образом,

$$\dot v_i(\xi_j, t) = \frac{D v_i(\vec \xi, t)}{Dt} = \frac{\partial}{\partial t} v_i(\vec \xi, t) = \frac{\partial}{\partial t} v_i(\vec x, t)$$

Уравнения движения (7) запишем в рамках линейной теории упругости в виде

$$\frac{\partial \sigma_{ji}\left(\vec x, t\right)}{\partial x_j}
+
X_i \left(\vec x, t\right) = \rho\,
\frac{\partial^2 u_i\left(\vec x, t\right) }{\partial t^2}$$

## ⚠ Исправление: скорость материальной частицы (правка 30.10.2026)

$x_i(\vec\xi,t)$ — **начальная** координата (метка) частицы. Для одной и той же частицы она не меняется, поэтому её материальная производная равна нулю, а не $v_i$:

$$\frac{Dx_i}{Dt}=\frac{\partial x_i}{\partial t}+v_j\frac{\partial x_i}{\partial \xi_j}=0 .$$

Скорость — это производная **текущего** положения: $v_i=D\xi_i/Dt$. Подставляя $x_i=\xi_i-u_i(\vec\xi,t)$:

$$0=-\frac{\partial u_i}{\partial t}+v_i-v_j\frac{\partial u_i}{\partial \xi_j}\quad\Longrightarrow\quad
\boxed{\left(\delta_{ij}-\frac{\partial u_i}{\partial \xi_j}\right)v_j=\frac{\partial u_i}{\partial t}}$$

Прежняя запись $\frac{\partial u_i}{\partial \xi_j}v_j=-\frac{\partial u_i}{\partial t}$ теряет единичную матрицу. Для плоской волны она давала $v=-\omega/k$ (фазовую скорость) при любой, сколь угодно малой амплитуде и требовала обхода случая $\det=0$. С единичной матрицей $\det(I-\nabla u)\neq0$ при малых деформациях, особый случай исчезает, а при $A\to0$ получаем $v\approx\partial u/\partial t$, то есть Ламе.

В Maple-процедуре `calc_material_vel` нужно заменить
`LinearSolve(du_dξ, minus_du_dt)` → `LinearSolve(IdentityMatrix(3) - du_dξ, du_dt)`, где `du_dt` — вектор $+\partial u_i/\partial t$. Выводы Maple в ячейках ниже посчитаны **старой** версией и подлежат пересчёту; проверка в Sage приведена в разделе «Проверки» в конце.

# Проверки (Sage) и решение во втором порядке
Ниже — независимая проверка в Sage. После каждого шага печатается `OK`/`FAIL`. Рассматривается одномерная задача (плоская волна вдоль $\xi$), $\theta=k\xi+\omega t+\delta$.

```python
var('xi t A k omega delta lam mu rho0 X0', domain='real')
assume(k>0, omega>0, rho0>0)
def check(name, expr):
    ok = bool(simplify(expr)==0)
    print(('OK   ' if ok else 'FAIL ')+name)
    return ok
th = k*xi + omega*t + delta
```

## 1. Скорость частицы — проверка на траектории
Частица с меткой $X_0$ находится в точке $\xi(t)$, где $\xi-u(\xi,t)=X_0$. Дифференцируя по $t$ неявно: $\dot\xi(1-u_\xi)=u_t$, т.е. $v=u_t/(1-u_\xi)$ — ровно формула с единичной матрицей. Старая формула $-u_t/u_\xi$ даёт $-\omega/k$.

```python
u = A*cos(th)
v_new = diff(u,t)/(1-diff(u,xi))
v_old = -diff(u,t)/diff(u,xi)
# неявное дифференцирование уравнения траектории  xi - u(xi,t) = X0
xs = function('xs')(t)
traj = xs - u.subs(xi=xs) - X0
dxs = solve(diff(traj,t)==0, diff(xs,t))[0].rhs()
check('v_new совпадает со скоростью вдоль траектории', dxs - v_new.subs(xi=xs))
print('v_old =', v_old.simplify_full(), ' (фазовая скорость, не зависит от A)')
# численно
pts = dict(A=0.05,k=2,omega=3,delta=0.3)
f = lambda s,tt: s - u.subs(pts).subs(xi=s,t=tt) - 0.4
for tt in [0.1,0.7,1.3]:
    s0 = find_root(lambda s: f(s,tt), -1, 2); h=1e-5
    vt = (find_root(lambda s: f(s,tt+h),-1,2)-find_root(lambda s: f(s,tt-h),-1,2))/(2*h)
    print(tt, 'траектория %.6f  v_new %.6f  v_old %.6f' % (vt, v_new.subs(pts).subs(xi=s0,t=tt), v_old.subs(pts).subs(xi=s0,t=tt)))
```

## 2. Продольная волна: непрерывность и невязка уравнения движения
$\rho=\rho_0(1-u_\xi)$ (якобиан), Альманси $\eta=u_\xi-\tfrac12u_\xi^2$, $\sigma=(\lambda+2\mu)\eta$.
Невязка $R=\rho(v_t+vv_\xi)-\sigma_\xi$ раскладывается по $A$; при $\omega^2=(\lambda+2\mu)k^2/\rho_0$ член $O(A)$ должен исчезнуть (Ламе).

```python
ux = diff(u,xi)
rho = rho0*(1-ux)
check('уравнение непрерывности', diff(rho,t)+diff(rho*v_new,xi))
sig = (lam+2*mu)*(ux-ux^2/2)
R = rho*(diff(v_new,t)+v_new*diff(v_new,xi)) - diff(sig,xi)
disp_L = {omega: k*sqrt((lam+2*mu)/rho0)}
R1 = diff(R,A).subs(A=0).subs(disp_L)
R2 = (diff(R,A,2)/2).subs(A=0).subs(disp_L)
check('порядок A: уравнение Ламе выполнено', R1)
check('порядок A^2: R2 = 3/2 (lam+2mu) k^3 sin(2θ)', R2 - 3/2*(lam+2*mu)*k^3*sin(2*th))
```

## 3. Поперечная волна (смещение по $Z$, распространение по $X$)
$\det(I-\nabla u)=1$, $v=(0,0,W_t)$ точно. Уравнение по $Z$ — точно Ламе, по $X$ — квадратичная продольная сила.

```python
W = A*cos(th); Wx = diff(W,xi)
G = matrix(SR,[[0,0,0],[0,0,0],[Wx,0,0]])
I3 = identity_matrix(SR,3)
check('det(I - grad u) = 1', (I3-G).det()-1)
vT = (I3-G).inverse()*vector(SR,[0,0,diff(W,t)])
check('v = (0,0,W_t)', (vT - vector(SR,[0,0,diff(W,t)])).norm()^2)
eta = (G+G.transpose()-G.transpose()*G)/2
sigT = lam*eta.trace()*I3 + 2*mu*eta
disp_T = {omega: k*sqrt(mu/rho0)}
check('уравнение по Z: Ламе точно', (rho0*diff(W,t,2)-diff(sigT[2,0],xi)).subs(disp_T))
F1 = diff(sigT[0,0],xi)
check('продольная сила F_X = -(lam+2mu)/2 k^3 A^2 sin(2θ)', F1 + (lam+2*mu)/2*k^3*A^2*sin(2*th))
```

## 4. Решение во втором порядке

**Переход к лагранжевой форме.** С правильной скоростью эйлерово уравнение эквивалентно лагранжеву. Для одномерного движения $\xi=X+U(X,t)$, $\partial\xi/\partial X=1+U_X$, $\eta_{Almansi}=U_X/(1+U_X)-\tfrac12 U_X^2/(1+U_X)^2$, и с точностью до $O(U_X^2)$

$$U_{tt}=c_L^2\left(1-2\beta\,U_X\right)U_{XX},\qquad c_L^2=\frac{\lambda+2\mu}{\rho_0},\qquad \beta=\tfrac32 .$$

(Коэффициент $\beta$ — следствие выбранного определяющего соотношения: линейный закон Гука по тензору Альманси. Для пятиконстантной теории Мурнагана $\beta$ включает модули третьего порядка; это и есть «параметр нелинейности» нелинейной акустики.)

**Регулярное разложение.** Граничное возбуждение $U(0,t)=A\sin\omega t$. Ищем $U=U_1+U_2$, $U_1=A\sin(\omega t-kX)$:

$$U_{2,tt}-c^2U_{2,XX}=-2\beta c^2 U_{1,X}U_{1,XX}=-\beta c^2k^3A^2\sin2(\omega t-kX).$$

Правая часть резонансна (бежит с той же скоростью), поэтому решение с $U_2(0,t)=0$ — секулярное:

$$\boxed{U_2=-\frac{\beta k^2A^2}{4}\,X\cos2(\omega t-kX)}$$

Амплитуда второй гармоники растёт линейно по пройденному пути: $|U_2|/|U_1|=\beta k^2 A X/4=\beta kM X/4$, где $M=kA$ — акустическое число Маха (амплитуда деформации).

**Расстояние образования разрыва.** Разложение справедливо при $X\ll x_s$:

$$x_s=\frac{1}{\beta k^2A}=\frac{1}{\beta k\,\varepsilon_0},\qquad \varepsilon_0=kA .$$

Точное простоволновое (риманово) решение $U_X=f\!\left(t-X/c(U_X)\right)$, $c(U_X)=c\sqrt{1-2\beta U_X}$, на расстоянии $x_s$ даёт опрокидывание профиля — образование ударной волны. При $X<x_s$ решение задаётся рядом Фубини: амплитуды гармоник деформации $\varepsilon_n=\dfrac{2\varepsilon_0}{n\sigma}J_n(n\sigma)$, $\sigma=X/x_s$.

```python
var('Xc c beta_', domain='real')
psi = omega*t - k*Xc
U1 = A*sin(psi)
U2 = -beta_*k^2*A^2/4*Xc*cos(2*psi)
lhs = diff(U2,t,2) - c^2*diff(U2,Xc,2)
rhs = -2*beta_*c^2*diff(U1,Xc)*diff(U1,Xc,2)
check('U2 решает уравнение второго порядка', (lhs-rhs).subs(omega=c*k))
check('U2(0,t)=0', U2.subs(Xc=0))
# Лагранжев коэффициент beta из эйлеровой невязки: 2*beta*c^2*rho0*k^3 A^2 sinθcosθ = R2*A^2
check('beta = 3/2 согласован с эйлеровой невязкой', 2*(3/2)*(lam+2*mu)*k^3*sin(th)*cos(th) - 3/2*(lam+2*mu)*k^3*sin(2*th))
```

## 5. Численная проверка: конечные разности
Решаем $U_{tt}=c^2(1-2\beta U_X)U_{XX}$ с $U(0,t)=A\sin\omega t$ и сравниваем амплитуду второй гармоники с $\beta k^2A^2X/4$, а затем строим профиль при приближении к $x_s$.

```python
import numpy as np
import matplotlib.pyplot as plt
def run(A=2e-3, beta=1.5, c=1.0, k=2*np.pi, L=6.0, N=3000, Tfrac=0.9):
    w=c*k; dx=L/N; dt=0.4*dx/c; x=np.arange(N+1)*dx
    U=np.zeros(N+1); Up=U.copy(); nt=int(Tfrac*L/c/dt); rec=[]; snap=None
    probe=int(3.0/dx)
    for n in range(nt):
        tn=n*dt
        Ux=(U[2:]-U[:-2])/(2*dx); Uxx=(U[2:]-2*U[1:-1]+U[:-2])/dx**2
        Un=U.copy(); Un[1:-1]=2*U[1:-1]-Up[1:-1]+dt**2*c**2*(1-2*beta*Ux)*Uxx
        Un[0]=A*np.sin(w*(tn+dt)); Un[-1]=0; Up,U=U,Un
        if tn>nt*dt-2*2*np.pi/w: rec.append((tn,U[probe]))
    rec=np.array(rec); tt,s=rec[:,0],rec[:,1]
    a=lambda m: 2*np.hypot(np.mean(s*np.sin(m*w*tt)),np.mean(s*np.cos(m*w*tt)))
    X=x[probe]
    print('X=%.2f  A1=%.3e  A2 числ=%.3e  теор βk²A²X/4=%.3e  x_s=%.2f'%(X,a(1),a(2),beta*k**2*A**2*X/4,1/(beta*k**2*A)))
    return x,U
x,U=run()
x2,U2n=run(A=6e-3)   # x_s ≈ 2.8: профиль деформации укручается
plt.plot(x2[1:-1],-(U2n[2:]-U2n[:-2])/(x2[2]-x2[0]),lw=0.8); plt.xlabel('X'); plt.ylabel('-U_X'); plt.title('укручение профиля деформации'); plt.show()
```

## Итог
1. Скорость частицы в описании Эйлера: $(I-\nabla u)\,v=\partial_t u$ (исправлено).
2. Уравнение движения в координатах Эйлера нелинейно; в первом порядке оно сводится к Ламе, синусоида — решение первого порядка.
3. Во втором порядке продольная волна генерирует вторую гармонику с амплитудой $\beta k^2A^2X/4$, растущей по пути; $\beta=3/2$ для закона Гука по Альманси.
4. Поперечная волна порождает продольную силу $-(\lambda+2\mu)k^3A^2\sin2\theta/2$ (вынужденная продольная вторая гармоника, несинхронная, т.к. $c_T\neq c_L$ — растёт ограниченно).
5. Расстояние образования разрыва $x_s=1/(\beta k^2 A)$.
